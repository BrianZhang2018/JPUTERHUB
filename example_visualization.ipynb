# Matplotlib Visualization Examples
This notebook demonstrates 4 common chart types using `matplotlib` and `numpy`.

```python
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['figure.dpi'] = 100

print('matplotlib and numpy loaded successfully!')
```

## Chart 1: Line Chart — Sine and Cosine Waves

```python
x = np.linspace(0, 2 * np.pi, 300)
y_sin = np.sin(x)
y_cos = np.cos(x)

plt.figure()
plt.plot(x, y_sin, label='sin(x)', color='blue', linewidth=2)
plt.plot(x, y_cos, label='cos(x)', color='orange', linewidth=2, linestyle='--')
plt.title('Sine and Cosine Waves')
plt.xlabel('x (radians)')
plt.ylabel('Amplitude')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Chart 2: Bar Chart — Monthly Sales Data

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales  = [12400, 15300, 11800, 17200, 19500, 22100]
colors = ['#4C72B0', '#DD8452', '#55A868', '#C44E52', '#8172B2', '#937860']

plt.figure()
bars = plt.bar(months, sales, color=colors, edgecolor='white', linewidth=0.8)
for bar, value in zip(bars, sales):
    plt.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 200,
             f'${value:,}', ha='center', va='bottom', fontsize=9)
plt.title('Monthly Sales — H1 2025')
plt.xlabel('Month')
plt.ylabel('Sales (USD)')
plt.ylim(0, 25000)
plt.grid(axis='y', alpha=0.4)
plt.tight_layout()
plt.show()
```

## Chart 3: Scatter Plot — Correlation Between Two Variables

```python
np.random.seed(42)
x = np.random.randn(200)
y = 2.5 * x + np.random.randn(200) * 0.8
colors = np.sqrt(x**2 + y**2)

plt.figure()
scatter = plt.scatter(x, y, c=colors, cmap='viridis', alpha=0.7, edgecolors='white', linewidths=0.4, s=60)
plt.colorbar(scatter, label='Distance from origin')
plt.title('Scatter Plot — Correlated Variables')
plt.xlabel('Variable X')
plt.ylabel('Variable Y')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Chart 4: Histogram — Normal Distribution

```python
np.random.seed(0)
data = np.random.normal(loc=0, scale=1, size=1000)

x_curve = np.linspace(-4, 4, 300)
y_curve = (1 / np.sqrt(2 * np.pi)) * np.exp(-0.5 * x_curve**2)

plt.figure()
plt.hist(data, bins=40, density=True, color='steelblue', edgecolor='white', alpha=0.7, label='Sampled data')
plt.plot(x_curve, y_curve, color='crimson', linewidth=2, label='Normal PDF')
plt.title('Histogram — Normal Distribution (n=1000)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
